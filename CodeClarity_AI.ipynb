```python
sample_code = """
def calculate_average(numbers):
    total = sum(numbers)
    return total / len(numbers)

numbers = []
print(calculate_average(numbers))
"""

print("========== CODECLARITY AI ==========\n")

print("===== BUG EXPLANATION =====\n")

print("Bug:")
print("The program tries to divide by zero.")

print("\nError Type:")
print("Runtime Error")

print("\nWhy it happens:")
print("The 'numbers' list is empty. Therefore, len(numbers) is 0.")
print("The expression total / len(numbers) becomes 0 / 0,")
print("which causes a ZeroDivisionError.")

print("\nHow to fix it:")
print("Check whether the list is empty before calculating the average.")

print("\nCorrected Code:")
print("""
def calculate_average(numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

numbers = []
print(calculate_average(numbers))
""")


print("\n===== CODE IMPROVEMENT =====\n")

print("1. Readability:")
print("The function name 'calculate_average' is clear and meaningful.")

print("\n2. Structure:")
print("The calculation is correctly placed inside a function.")

print("\n3. Efficiency:")
print("sum() and len() are efficient for calculating the average.")

print("\n4. Error Handling:")
print("The function should handle an empty list.")

print("\n5. Security:")
print("There are no major security issues in this code.")

print("\n6. Best Practice:")
print("Use an empty-list check before performing division.")

print("\n7. Improved Code:")
print("""
def calculate_average(numbers):
    if not numbers:
        return 0
    return sum(numbers) / len(numbers)

numbers = []
print(calculate_average(numbers))
""")
```
